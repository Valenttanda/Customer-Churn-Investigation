## Load datasets

## Checking datasets

```python
# Missing values checking
def check_missing_values(df):
  missing_values = df.isna().sum()
  missing_values = missing_values[missing_values > 0]
  missing_percentage = missing_values / len(df) * 100
  missing_percentage = missing_percentage[missing_percentage > 0].round(2).astype(str) + ' %'
  missing_display = pd.DataFrame({
      	'Missing Values': missing_values,
				'Missing Percentage': missing_percentage
	})
  return missing_display
  
# Duplicate values checking
def check_duplicate_values(df):
	duplicate_values = df.duplicated().sum()
	return duplicate_values
```

## Handling datasets
### train_df dataset

```python
# Drop the missing values in the train_df
train_df_cleaned = train_df.dropna()
check_missing_values(train_df_cleaned) if check_missing_values(train_df_cleaned).shape[0] > 0 else print("No missing values found after dropping.")
```

```python
# Change all 'float64' columns to 'int64' in the train_df_cleaned
for col in train_df_cleaned.select_dtypes(include=['float64']).columns:
		train_df_cleaned[col] = train_df_cleaned[col].astype('int64')
train_df_cleaned.info()
```

```python
# Merge train_df_cleaned and test_df on 'customerID'
merged_df = pd.concat([train_df_cleaned, test_df], axis=0, ignore_index=True)
merged_df.info()
```

```python
# Save the cleaned and merged dataset to a new CSV file
merged_df.to_csv('data\customer_churn_dataset-cleaned.csv', index=False)
```

## EDA: Business Questions

### 1. How much total revenue is lost (Revenue at Risk) due to churn, and which subscription plans result in the greatest financial loss?

```python
first_question = merged_df.groupby(['Churn','Subscription Type'], as_index=False).agg(
	customers_count=('CustomerID','count'),
	total_revenue=('Total Spend','sum'),
	spend_average=('Total Spend','mean')
)
first_question
```

### 2. How significant is the difference in churn rates between Monthly, Quarterly, and Annual contracts?

```python
second_question = merged_df.groupby('Contract Length', as_index=False).agg(
	customers_total=('CustomerID','count'),
	churned_customers=('Churn','sum'),
	churn_rate=('Churn','mean'),
	total_spend=('Total Spend','sum')
)
second_question
```

### 3. At what threshold does the frequency of customer support calls cause customers to stop using the service?

```python
third_question = merged_df.groupby('Support Calls', as_index=False).agg(
	customers_total=('CustomerID','count'),
	churn_rate=('Churn','mean'),
	spend_average=('Total Spend','mean')
)
third_question
```

### 4. When does a payment delay go from being a simple oversight to a clear sign of churn?

```python
# 1. Daily Payment Delay
fourth_question_1 = merged_df.groupby('Payment Delay', as_index=False).agg(
	customers_total=('CustomerID','count'),
	churn_rate=('Churn','mean'),
	revenue_at_risk=('Total Spend','sum')
)
fourth_question_1
```

```python
# 2. Weekly Payment Delay
merged_df['Payment Delay (weeks)'] = pd.cut(
	merged_df['Payment Delay'], 
	bins=[-1, 0, 7, 14, 21, 30], 
	labels=['0 days', '1 week', '2 weeks', '3 weeks', '>3 weeks']
)

fourth_question_2 = merged_df.groupby('Payment Delay (weeks)', observed=False, as_index=False).agg(
	customers_total=('CustomerID','count'),
	churn_rate=('Churn','mean'),
	revenue_at_risk=('Total Spend','sum')
)
fourth_question_2
```

### 5. What are the customer segment profiles based on a combination of spending (Total Spend) and activity (Usage Frequency)?

```python
merged_df['Spend Tier'] = pd.qcut(merged_df['Total Spend'], q=2, labels=['Low Spend', 'High Spend'])
merged_df['Usage Tier'] = pd.qcut(merged_df['Usage Frequency'], q=2, labels=['Low Usage', 'High Usage'])

fifth_question = merged_df.groupby(['Spend Tier','Usage Tier'], observed=False, as_index=False).agg(
	customers_total=('CustomerID','count'),
	churn_rate=('Churn','mean'),
	average_calls=('Support Calls','mean'),
	spend_total=('Total Spend','sum')
)
fifth_question
```

### 6. Does a decline in activity (low usage frequency and a long time since the last interaction) help detect silent churn earlier?

```python
merged_df['Interaction Recency'] = pd.cut(
	merged_df['Last Interaction'], 
	bins=[0, 10, 20, 30], 
	labels=['Active (1-10 days)', 'Passive (11-20 days)', 'Inactive (>20 days)']
)

sixth_question = merged_df.groupby(['Usage Tier','Interaction Recency'], observed=False, as_index=False).agg(
	customers_total=('CustomerID','count'),
	churn_rate=('Churn','mean'),
	average_tenure=('Tenure','mean')
)
sixth_question
```

### 7. During the subscription period (Tenure), in which month are customers most likely to cancel their subscription?

```python
merged_df['Tenure Phase'] = pd.cut(
	merged_df['Tenure'],
	bins=[0, 12, 24, 36, 48, 60],
	labels=['0-12 months', '13-24 months', '25-36 months', '37-48 months', '49-60 months']
)

seventh_question = merged_df.groupby(['Tenure Phase','Subscription Type'], observed=False, as_index=False).agg(
	customers_total=('CustomerID','count'),
	churn_rate=('Churn','mean'),
	average_usage=('Usage Frequency','mean')
)
seventh_question
```

### 8. Do certain age groups or genders have a significantly higher churn rate?

```python
merged_df['Age Group'] =pd.cut(
	merged_df['Age'],
	bins=[17, 30, 40, 50, 60, 70],
	labels=['18-30 years', '31-40 years', '41-50 years', '51-60 years', '>60 years']
)

eighth_question = merged_df.groupby(['Age Group','Gender'], observed=False, as_index=False).agg(
	customers_total=('CustomerID','count'),
	churn_rate=('Churn','mean'),
	average_calls=('Support Calls','mean'),
 	average_usage=('Usage Frequency','mean')
)
eighth_question
```

## EDA: Visualization

```python
import plotly.express as px
import plotly.graph_objects as go

COLOR_SAFE = "#2E86AB"     # Blue (Retained)
COLOR_RISK = "#E63946"     # Red (Churned)
COLOR_NEUTRAL = "#457B9D"  # Blue-gray (Neutral)

# --- Q1: Grouped Bar Chart (Revenue at Risk) ---
def plot_q1(q1_df):
    df_plot = q1_df.copy()
    df_plot['Status'] = df_plot['Churn'].map({0: 'Retained', 1: 'Churned'})
    fig = px.bar(
        df_plot, x='Subscription Type', y='total_revenue', color='Status',
        barmode='group', text_auto='.3s',
        color_discrete_map={'Retained': COLOR_SAFE, 'Churned': COLOR_RISK},
        title='Question 1: Retained vs Churned Revenue per Subscription Type'
    )
    fig.update_layout(yaxis_title='Revenue ($)', xaxis_title='Subscription Type')
    return fig

# --- Q2: Horizontal Bar Chart (Contract Duration) ---
def plot_q2(q2_df):
    df_plot = q2_df.sort_values('churn_rate', ascending=True).copy()
    df_plot['churn_pct'] = df_plot['churn_rate'] * 100
    fig = px.bar(
        df_plot, x='churn_pct', y='Contract Length', orientation='h',
        text=df_plot['churn_pct'].apply(lambda x: f'{x:.1f}%'),
        color='Contract Length',
        color_discrete_map={'Monthly': COLOR_RISK, 'Annual': COLOR_SAFE, 'Quarterly': COLOR_NEUTRAL},
        title='Question 2: Churn Rate by Contract Duration (%)'
    )
    fig.update_layout(xaxis_title='Churn Rate (%)', yaxis_title='Contract Duration', showlegend=False, xaxis_range=[0, 105])
    return fig

# --- Q3: Line Chart + Threshold (Support Calls) ---
def plot_q3(q3_df):
    df_plot = q3_df.copy()
    df_plot['churn_pct'] = df_plot['churn_rate'] * 100
    fig = px.line(
        df_plot, x='Support Calls', y='churn_pct', markers=True,
        title='Question 3: Spike in Churn Rate Based on Support Call Frequency'
    )
    fig.update_traces(line_color='#1D3557', line_width=3, marker_size=8)
    # Add vertical line tipping point at 5 calls
    fig.add_vline(x=5, line_dash="dash", line_color=COLOR_RISK, annotation_text="Critical Threshold (5 Calls)")
    fig.update_layout(xaxis_title='Number of Support Calls', yaxis_title='Churn Rate (%)', yaxis_range=[0, 105])
    return fig

# --- Q4: Bar Chart (Payment Delay Bucket) ---
def plot_q4(q4_df):
    df_plot = q4_df.copy()
    df_plot['churn_pct'] = df_plot['churn_rate'] * 100
    fig = px.bar(
        df_plot, x='Payment Delay (weeks)', y='churn_pct',
        text=df_plot['churn_pct'].apply(lambda x: f'{x:.1f}%'),
        color='churn_pct', color_continuous_scale='YlOrRd',
        title='Question 4: Escalation of Churn Risk Based on Payment Delays (in weeks)'
    )
    fig.update_layout(xaxis_title='Payment Delay Category', yaxis_title='Churn Rate (%)', coloraxis_showscale=False, yaxis_range=[0, 105])
    return fig

# --- Q5: Matrix Heatmap (Spend vs Usage Segmentation) ---
def plot_q5(q5_df):
    pivot_df = q5_df.pivot(index='Spend Tier', columns='Usage Tier', values='churn_rate') * 100
    fig = px.imshow(
        pivot_df, text_auto='.1f', color_continuous_scale='YlOrRd',
        labels=dict(x="Usage Tier", y="Spend Tier", color="Churn Rate (%)"),
        title='Question 5: Customers Segmentation Matrix — Churn Rate (%)'
    )
    return fig

# --- Q6: Grouped Bar Chart (Silent Churn) ---
def plot_q6(q6_df):
    df_plot = q6_df.copy()
    df_plot['churn_pct'] = df_plot['churn_rate'] * 100
    fig = px.bar(
        df_plot, x='Interaction Recency', y='churn_pct', color='Usage Tier',
        barmode='group', text=df_plot['churn_pct'].apply(lambda x: f'{x:.1f}%'),
        color_discrete_map={'Low Usage': COLOR_RISK, 'High Usage': COLOR_SAFE},
        title='Question 6: Silent Churn Detection (Usage Frequency vs Last Interaction)'
    )
    fig.update_layout(xaxis_title='Last Interaction Time', yaxis_title='Churn Rate (%)', yaxis_range=[0, 80])
    return fig

# --- Q7: Multi-Line Chart (Tenure Phase) ---
def plot_q7(q7_df):
    df_plot = q7_df.copy()
    df_plot['churn_pct'] = df_plot['churn_rate'] * 100
    fig = px.line(
        df_plot, x='Tenure Phase', y='churn_pct', color='Subscription Type', markers=True,
        title='Question 7: Trend of Churn Rate Throughout Customer Tenure'
    )
    fig.update_layout(xaxis_title='Tenure Phase', yaxis_title='Churn Rate (%)', yaxis_range=[40, 75])
    return fig

# --- Q8: Grouped Bar Chart (Age & Gender Demography) ---
def plot_q8(q8_df):
    df_plot = q8_df.copy()
    df_plot['churn_pct'] = df_plot['churn_rate'] * 100
    fig = px.bar(
        df_plot, x='Age Group', y='churn_pct', color='Gender',
        barmode='group', text=df_plot['churn_pct'].apply(lambda x: f'{x:.1f}%'),
        color_discrete_map={'Female': '#E76F51', 'Male': '#2A9D8F'},
        title='Question 8: Comparison of Churn Rate by Age Group and Gender'
    )
    fig.update_layout(xaxis_title='Age Group', yaxis_title='Churn Rate (%)', yaxis_range=[0, 105])
    return fig
```
